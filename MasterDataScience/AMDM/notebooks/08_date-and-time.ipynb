## 1. Calendar Module

*Difficulty: Easy*

```python
import calendar

month, day, year = map(int, input().split())

day_index = calendar.weekday(year, month, day)

day_name = calendar.day_name[day_index]

print(day_name.upper())
```

**Explanation:** The program reads a date as three integers, in the order month, day, year, and unpacks them into `month`, `day` and `year`. `calendar.weekday(year, month, day)` returns the day of the week as a number, where 0 is Monday and 6 is Sunday. That number is used as an index into `calendar.day_name`, a list of the weekday names, to get the name of the day. `.upper()` converts it to capitals, as the exercise requires, and the result is printed. Note that `calendar.weekday` takes its arguments in a different order from the input, so the order in the call matters.

## 2. Time Delta

*Difficulty: Medium*

```python
import math
import os
import random
import re
import sys

from datetime import datetime

def time_delta(t1, t2):
    formato = "%a %d %b %Y %H:%M:%S %z"
    dt1 = datetime.strptime(t1, formato)
    dt2 = datetime.strptime(t2, formato)

    return str(int(abs((dt1 - dt2).total_seconds())))


if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    t = int(input())

    for t_itr in range(t):
        t1 = input()

        t2 = input()

        delta = time_delta(t1, t2)

        fptr.write(delta + '\n')

    fptr.close()
```

**Explanation:** The function `time_delta(t1, t2)` parses two timestamps given in the format `Sun 10 May 2015 13:54:36 -0700` using `datetime.strptime` with the format string `"%a %d %b %Y %H:%M:%S %z"`. The `%z` directive reads the UTC offset, so each result is a timezone-aware datetime. Subtracting them gives a `timedelta` that already accounts for the different offsets, so no manual conversion is needed. `.total_seconds()` turns the difference into seconds, `abs()` removes the sign so the order of the two timestamps doesn't matter, and `int()` and `str()` produce the integer text that the main block writes to the output file, one line per test case.
