## 1. Zipped!

*Difficulty: Easy*

```python
c, r = map(int, input().split())

repo = []
for _ in range(r):
    valori = list(map(float, input().split()))
    repo.append(valori)

for scores in zip(*repo):
    print(sum(scores) / len(scores))
```

**Explanation:** The first line gives `c`, the number of students, and `r`, the number of subjects. The program then reads `r` lines of marks, one per subject, and stores each as a list of floats in `repo`. `zip(*repo)` unpacks those lists and pairs up the values in the same position, so each tuple `scores` holds all the marks of one student across the subjects. For each student, `sum(scores) / len(scores)` computes the average, and it is printed on its own line, in the order the students appear in the input.

## 4. Athlete Sort

*Difficulty: Medium*

```python
import math
import os
import random
import re
import sys

if __name__ == '__main__':
    first_multiple_input = input().rstrip().split()

    n = int(first_multiple_input[0])

    m = int(first_multiple_input[1])

    arr = []

    for _ in range(n):
        arr.append(list(map(int, input().rstrip().split())))

    k = int(input().strip())

arr.sort(key=lambda n: n[k])

for a in arr:
    print(*(a))
```

**Explanation:** The program reads `n` and `m`, the number of athletes and the number of attributes, then reads `n` rows of integers into the list of lists `arr`, and finally the index `k` of the attribute to sort by. `arr.sort(key=lambda n: n[k])` sorts the rows in place, using the value at position `k` of each row as the sort key. Python's sort is stable, so athletes with the same value keep the order they had in the input, which is what the exercise requires for ties. Each sorted row is then printed on its own line with `print(*(a))`, which unpacks the numbers separated by spaces.

## 6. ginortS

*Difficulty: Medium*

```python
nome = input()

low = []
up = []
odd_num = []
even_num = []

for char in nome:
    if char.islower() == True:
        low.append(char)
    elif char.isupper() == True:
        up.append(char)
    elif int(char) % 2 != 0:
        odd_num.append(char)
    else:
        even_num.append(char)
        
tutto = sorted(low) + sorted(up) + sorted(odd_num) + sorted(even_num)
print("".join(tutto))
```

**Explanation:** The program reads a string and sorts it into four groups by character type: lowercase letters, uppercase letters, odd digits and even digits. It goes through the characters one at a time, and an `if/elif/else` chain appends each one to the matching list. Digits are tested with `int(char) % 2`, so the remainder separates the odd ones from the even ones. Each of the four lists is sorted on its own and the lists are concatenated in the order the exercise requires: lowercase, uppercase, odd digits, even digits. `"".join(tutto)` turns the final list of characters back into a single string, which is printed.
