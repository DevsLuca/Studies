## 1. collections.Counter()

*Difficulty: Easy*

```python
from collections import Counter

num_shoes = int(input())
store = Counter(map(int, input().split()))
clients = int(input())

profit = 0

for _ in range(clients):
    size, price = map(int, input().split())

    if store[size] > 0:
        profit += price
        store[size] -= 1

print(profit)
```

**Explanation:** The program reads the shop's stock as a list of shoe sizes and stores it in a `Counter`, a dictionary that maps each size to the number of pairs available. For each customer it then reads the size they want and the price they offer. If the shop still has at least one pair of that size (`store[size] > 0`), the price is added to `profit` and the count for that size goes down by one. A `Counter` returns 0 for sizes it has never seen, so no extra check is needed for sizes missing from the stock. The final `profit` is printed.

## 2. DefaultDict Tutorial

*Difficulty: Easy*

```python
from collections import defaultdict

n, m = map(int, input().split())
d = defaultdict(list)

for i in range(n):
    word = input()
    d[word].append(i+1)

for _ in range(m):
    word = input()
    if word in d:
        print(*(d[word]))
    else:
        print(-1)
```

**Explanation:** The program reads `n` and `m`, the sizes of two groups of words, and stores the first group in a `defaultdict(list)`. Each word becomes a key and its value is the list of positions where it appeared, counting from 1 (`i+1`), so a repeated word collects several indices without any initialisation code. For each of the `m` words in the second group, the program checks whether it is a key with `word in d`. If it is, `print(*(d[word]))` prints all its positions on one line, and if not it prints `-1`.

## 3. Collections.namedtuple()

*Difficulty: Easy*

```python
students = int(input())
columns = input().split()
col_index = columns.index("MARKS")

marks = []
for student in range(students):
    row = input().split()
    mark = int(row[col_index])
    marks.append(mark)
    
average = sum(marks) / students
print(f"{average:.2f}")
```

**Explanation:** The program reads the number of students and then the line of column names, which can come in any order. `columns.index("MARKS")` finds the position of the `MARKS` column, so the code works wherever that column appears. For each student, it splits the input row, takes the value at that position, converts it to an integer and appends it to the list `marks`. The average is the sum of the marks divided by the number of students, printed with two decimal places using the format `:.2f`.

## 4. Collections.OrderedDict()

*Difficulty: Easy*

```python
from collections import OrderedDict
shopping = OrderedDict()

n = int(input())

for _ in range(n):
    item_name, net_price = input().rsplit(' ', 1)
    if item_name in shopping:
        shopping[item_name] += int(net_price)
    else:
        shopping[item_name] = int(net_price)

for item_name, net_price in shopping.items():
    print(f"{item_name} {net_price}")
```

**Explanation:** The program reads `n` purchase lines, each made of an item name followed by its price, and totals the price of every item in an `OrderedDict`. Each line is split with `rsplit(' ', 1)`, which cuts only at the last space, so names made of several words stay intact and the final word becomes the price. If the item is already in the dictionary its price is added to the running total, and otherwise a new entry is created. An `OrderedDict` remembers the order in which keys were first inserted, so printing it gives the items in order of first appearance, each with its net price.

## 5. Word Order

*Difficulty: Medium*

```python
from collections import defaultdict

words = defaultdict(int)

n = int(input())

for _ in range(n):
    word = input()
    words[word] += 1

print(len(words))
print(*(words.values()))
```

**Explanation:** The program reads `n` words and counts how many times each one appears, using a `defaultdict(int)`. A word that is not yet in the dictionary starts at 0, so `words[word] += 1` works without checking for it first. The number of distinct words is the number of keys, so `len(words)` is printed first. `print(*(words.values()))` then prints the counts on one line, in the order the words first appeared, because dictionaries keep the order in which keys were inserted.

## 6. Collections.deque()

*Difficulty: Easy*

```python
from collections import deque

d = deque()
n = int(input())

for _ in range(n):
    parts = input().split()
    op = parts[0]
    
    if op == 'append':
        d.append(int(parts[1]))
    elif op == 'appendleft':
        d.appendleft(int(parts[1]))
    elif op == 'pop':
        d.pop()
    elif op == 'popleft':
        d.popleft()

print(*d)
```

**Explanation:** The program starts with an empty `deque`, a double-ended queue that can add and remove items efficiently at both ends. It reads `n` commands, splits each line into words, and uses the first word to decide which method to call: `append` and `appendleft` add the integer in the second word to the right or left end, and `pop` and `popleft` remove an item from the right or left end. An `if/elif` chain dispatches the command to the matching method. After all the commands, `print(*d)` prints the remaining elements on one line, from left to right.

## 7. Company Logo

*Difficulty: Medium*

```python
from collections import defaultdict

if __name__ == '__main__':
    s = input()
    d = defaultdict(int)

    for char in s:
        d[char] += 1
        
    sorted_d = dict(sorted(d.items()))

    n = 3
    top_items = []
    seen_keys = set()

    for _ in range(n):
        max_key = None
        max_val = 0  
        
        for key, val in sorted_d.items():
            if key not in seen_keys:
                if val > max_val:
                    max_val = val
                    max_key = key
        
        top_items.append((max_key, max_val))
        seen_keys.add(max_key)

    for char, count in top_items:
        print(f"{char} {count}")
```

**Explanation:** The program counts how many times each character appears in the string `s` with a `defaultdict(int)`, and then sorts the counts alphabetically by character with `sorted(d.items())`. It then picks the three most common characters one at a time: on each pass it scans the sorted dictionary for the largest count among the characters not yet chosen, which are tracked in the set `seen_keys`. Because the dictionary is already in alphabetical order and the comparison is a strict `val > max_val`, ties always go to the character that comes first in the alphabet, as the exercise requires. Each chosen character and its count is saved in `top_items` and printed on its own line.

## 8. Piling Up!

*Difficulty: Medium*

```python
t = int(input())

for _ in range(t):
    size = int(input())
    block = list(map(int, input().split()))
    L = 0
    R = len(block) - 1
    base_size = float("inf")
    check = "Yes"
    while L <= R:
        if block[L] >= block[R]:
            if block[L]>base_size:
                check = "No"
                break
            base_size = block[L]
            L +=1
        else:
            if block[R]>base_size:
                check = "No"
                break
            base_size = block[R]
            R -= 1
    print(check)
```

**Explanation:** For each test case the program reads the list of cube side lengths and keeps two pointers, `L` and `R`, at the left and right ends of the row. At each step it takes the larger of the two end cubes, since that is the best choice for building a pile that never gets wider going upward. If that cube is bigger than the one placed just before (`base_size`), the pile is impossible and the answer becomes `"No"` immediately. Otherwise the cube becomes the new `base_size`, the matching pointer moves inward, and the loop repeats until the pointers meet. If every cube fits, the answer stays `"Yes"`, and it is printed once per test case.
