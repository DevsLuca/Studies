## 1. Arrays

*Difficulty: Easy*

```python
import numpy

def arrays(arr):
    lista = []
    for i in range(len(arr)-1, -1, -1):
        lista.append(arr[i])
        
    lista = numpy.array(lista, float)
    return lista
    
    # complete this function
    # use numpy.array

arr = input().strip().split(' ')
result = arrays(arr)
print(result)
```

**Explanation:** The program reads one line of space-separated numbers and splits it into `arr`, a list of strings. The function `arrays(arr)` copies the elements into a new list `lista` in reverse order, using a loop that counts down from the last index to 0. It then calls `numpy.array(lista, float)`, which converts the list into a NumPy array and turns every string into a floating-point number at the same time. The resulting array is returned and printed, so the output shows the values in reverse order with decimals.

## 2. Shape and Reshape

*Difficulty: Easy*

```python
import numpy

a = list(map(int, input().split()))
a = numpy.array(a)
print(numpy.reshape(a, (3,3)))
```

**Explanation:** The program reads one line of nine space-separated integers and converts it into a list with `map(int, ...)`. `numpy.array(a)` turns that list into a one-dimensional NumPy array. `numpy.reshape(a, (3,3))` then rearranges the same nine values into a 3 x 3 matrix, filling it row by row in the original order. The reshaped array is printed, and the original data is not changed, because reshape returns a new view of it.

## 3. Transpose and Flatten

*Difficulty: Easy*

```python
import numpy

n, m = map(int, input().split())

rows = []
for _ in range(n):
    row_data = input().split()
    rows.append(row_data)

matrix = numpy.array(rows, int)

print(matrix.T)
print(matrix.flatten())
```

**Explanation:** The program reads `n` and `m`, the number of rows and columns, and then reads `n` lines of numbers, storing each one as a list of strings in `rows`. `numpy.array(rows, int)` builds a 2D array from this list and converts every string to an integer in the same step. `matrix.T` gives the transpose, in which rows and columns are swapped, and `matrix.flatten()` collapses the matrix into a one-dimensional array, reading it row by row. The two results are printed one after the other.

## 4. Concatenate

*Difficulty: Easy*

```python
import numpy

n, m, p = map(int, input().split())

A = []
for _ in range(n):
    A_data = input().split()
    A.append(A_data)
A = numpy.array(A, int)
B = []
for _ in range(m):
    B_data = input().split()
    B.append(B_data)
B = numpy.array(B, int)

print (numpy.concatenate((A, B), axis=0))
```

**Explanation:** The first line gives `n`, `m` and `p`: array `A` has `n` rows and array `B` has `m` rows, and both have `p` columns. The program reads `n` lines into `A` and `m` lines into `B`, and each is converted to an integer array with `numpy.array(..., int)`. `numpy.concatenate((A, B), axis=0)` joins them along the first axis, which stacks `B` underneath `A` and gives a single array with `n + m` rows. The `axis=0` argument is what selects row-wise stacking, and joining along the columns would require `axis=1`.

## 5. Zeros and Ones

*Difficulty: Easy*

```python
import numpy as np

shape = tuple(map(int, input().split()))

print(np.zeros(shape, dtype=int))
print(np.ones(shape, dtype=int))
```

**Explanation:** The program reads one line of space-separated integers and converts it into a tuple, `shape`, which describes the dimensions of the arrays to build. `np.zeros(shape, dtype=int)` creates an array of that shape filled with zeros, and `np.ones(shape, dtype=int)` creates one filled with ones. The `dtype=int` argument is needed because NumPy would otherwise produce floating-point values like `0.` and `1.`. Both arrays are printed one after the other, with the zeros first.

## 7. Array Mathematics

*Difficulty: Easy*

```python
import numpy as np

n, m = map(int, input().split())

A = np.array([input().split() for _ in range(n)], int)
B = np.array([input().split() for _ in range(n)], int)

print(A + B)
print(A - B)
print(A * B)
print(A // B)
print(A % B)
print(A ** B)
```

**Explanation:** The program reads `n` and `m`, the number of rows and columns, and then builds two integer arrays `A` and `B`, each from `n` lines of input, using a list comprehension inside `np.array(..., int)`. It then applies six arithmetic operators to the two arrays, and NumPy carries out each one element by element: addition, subtraction, multiplication, floor division (`//`), modulo (`%`) and power (`**`). No loops are needed, because the operators work on whole arrays at once, and the arrays must have the same shape. Each result is a new array, printed in the order the exercise requires.

## 8. Floor, Ceil and Rint

*Difficulty: Easy*

```python
import numpy as np

np.set_printoptions(legacy='1.13')

my_array = np.array(input().split(), float)

print(np.floor(my_array))
print(np.ceil(my_array))
print(np.rint(my_array))
```

**Explanation:** The program reads one line of space-separated numbers and converts it into a one-dimensional NumPy array of floats with `np.array(input().split(), float)`. It then applies three rounding functions to the whole array at once, with no loop. `np.floor` rounds each value down to the nearest integer, `np.ceil` rounds each one up, and `np.rint` rounds to the nearest integer, sending exact halves to the nearest even number. The line `np.set_printoptions(legacy='1.13')` makes NumPy print arrays in the older style that HackerRank's expected output uses.

## 9. Sum and Prod

*Difficulty: Easy*

```python
import numpy

n, m = map(int, input().split())

A = []
for _ in range(n):
    A_data = input().split()
    A.append(A_data)
A = numpy.array(A, int)

somma = numpy.sum(A, axis = 0) 
print(numpy.prod(somma, axis = 0))
```

**Explanation:** The program reads `n` and `m`, the number of rows and columns, then reads `n` lines of numbers and builds a 2D integer array `A`. `numpy.sum(A, axis=0)` adds the values down each column and returns a one-dimensional array with one total per column, stored in `somma`. `numpy.prod(somma, axis=0)` then multiplies those column totals together into a single number, which is printed. The order matters: the sum comes first and the product second, and `axis=0` is what makes the sum work column by column instead of over the whole array.

## 11. Mean, Var, and Std

*Difficulty: Easy*

```python
import numpy as np
#np.set_printoptions(legacy='1.13')

n, m = map(int, input().split())

arr = np.array([input().split() for _ in range(n)], int)

print(np.mean(arr, axis=1)) 
print(np.var(arr, axis=0))  
print(round(np.std(arr), 11))
```

**Explanation:** The program reads `n` and `m`, the number of rows and columns, and builds a 2D integer array `arr` from the next `n` lines. It then computes three statistics with different scopes. `np.mean(arr, axis=1)` gives the mean of each row, `np.var(arr, axis=0)` gives the variance of each column, and `np.std(arr)` gives the standard deviation of all the elements taken together. The last value is passed through `round(..., 11)` to print it with 11 decimal places, as the exercise requires.

## 12. Dot and Cross

*Difficulty: Easy*

```python
import numpy as np

n = int(input())

A = np.array([input().split() for _ in range(n)], int)
B = np.array([input().split() for _ in range(n)], int)

print(A @ B)
```

**Explanation:** The program reads `n` and builds two `n x n` integer arrays, `A` and `B`, from the next `2n` lines of input. The `@` operator computes their matrix product, in which each entry of the result is the dot product of a row of `A` with a column of `B`. For two-dimensional arrays this is the same as `np.dot(A, B)`, but the operator is shorter and reads more like the mathematical notation. The resulting `n x n` array is printed.

## 13. Inner and Outer

*Difficulty: Easy*

```python
import numpy as np

A = np.array([input().split()], int)
B = np.array([input().split()], int)

print(np.inner(A, B)[0,0])
print(np.outer(A, B))
```

**Explanation:** The program reads two lines of integers and stores each as a NumPy array `A` and `B`. The square brackets around `input().split()` give each array one row, so it has the shape `(1, n)`. `np.inner(A, B)` multiplies the two arrays element by element and adds the products, and because the result is a 1 x 1 array, `[0,0]` extracts the single number. `np.outer(A, B)` multiplies every element of `A` by every element of `B` and returns an `n x n` matrix, printed on the next line.

## 14. Polynomials

*Difficulty: Easy*

```python
import numpy as np

A = list(map(float, input().split()))

k = int(input())

print(np.polyval(A, k))
```

**Explanation:** The program reads one line of numbers, the coefficients of a polynomial, and converts them to a list of floats `A`. The coefficients are given from the highest power to the constant term, which is the order `np.polyval` expects. The second input line gives the integer `k`, the value at which the polynomial is evaluated. `np.polyval(A, k)` substitutes `k` into the polynomial and returns the result, so no loop or manual powers are needed.

## 15. Linear Algebra

*Difficulty: Easy*

```python
import numpy as np

n = int(input())

A = np.array([input().split() for _ in range(n)], float)

print(round(np.linalg.det(A), 2))
```

**Explanation:** The program reads `n`, the size of a square matrix, and builds an `n x n` array `A` of floats from the next `n` lines of input. `np.linalg.det(A)` computes the determinant, using the linear algebra routines in NumPy's `linalg` module. The result is a floating-point number with many decimals, so `round(..., 2)` trims it to two decimal places, as the exercise requires. The determinant is printed as a single number.
