## 1. XML 1 - Find the Score

*Difficulty: Easy*

```python
import sys
import xml.etree.ElementTree as etree

def get_attr_number(node):
    score = len(node.attrib)
    for child in node:
        score += get_attr_number(child)
    return score

if __name__ == '__main__':
    sys.stdin.readline()
    xml = sys.stdin.read()
    tree = etree.ElementTree(etree.fromstring(xml))
    root = tree.getroot()
    print(get_attr_number(root))
```

**Explanation:** The score of an XML document is the total number of attributes across all of its elements. The program skips the first input line, reads the rest as a single string, and parses it with `etree.fromstring` to get the root element. The recursive function `get_attr_number(node)` starts from `len(node.attrib)`, the number of attributes of the current element, and adds the same count for every child by calling itself on each one. Since the recursion visits the root and all of its descendants, the value returned for the root is the score of the whole document, which is then printed.

## 2. XML2 - Find the Maximum Depth

*Difficulty: Easy*

```python
import xml.etree.ElementTree as etree

maxdepth = 0
def depth(elem, level):
    global maxdepth
    if level == maxdepth:
        maxdepth += 1
    for child in elem:
        depth(child, level+1)

if __name__ == '__main__':
    n = int(input())
    xml = ""
    for i in range(n):
        xml =  xml + input() + "\n"
    tree = etree.ElementTree(etree.fromstring(xml))
    depth(tree.getroot(), -1)
    print(maxdepth)
```

**Explanation:** The program reads `n` lines of XML, joins them into one string, and parses it with `etree.fromstring` to get the root element. The recursive function `depth(elem, level)` walks the tree, calling itself on every child with `level + 1`. A global counter `maxdepth` grows by one each time the walk reaches a level equal to the current `maxdepth`, which only happens when it goes deeper than any level seen so far. The first call starts at level `-1`, so the root element counts as depth 0 and its children as depth 1, and `maxdepth` ends up equal to the number of levels below the root.
