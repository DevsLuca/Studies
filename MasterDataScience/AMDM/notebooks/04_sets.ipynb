## 1. Introduction to Sets

*Difficulty: Easy*

```python
def average(array):
    array = set(array)
    average = sum(array)/len(array)
    return round(average, 3)

if __name__ == '__main__':
    n = int(input())
    arr = list(map(int, input().split()))
    result = average(arr)
    print(result)
```

**Explanation:** The program reads `n` and then a line of plant heights, converted to a list of integers. The function `average(array)` first turns the list into a `set`, which removes duplicate heights so each distinct value counts only once. It then divides the sum of the remaining values by their number, `sum(array)/len(array)`, to get the average, and returns it rounded to three decimal places with `round(average, 3)`. Using a set for de-duplication is the key step, because averaging the original list would give a different result whenever heights repeat.

## 2. No Idea!

*Difficulty: Medium*

```python
_ = list(map(int, input().split()))
s = list(map(int, input().split()))
A = set(map(int, input().split()))
B = set(map(int, input().split()))

happy = 0
for i in s:
    if i in A:
        happy += 1
    if i in B:
        happy -= 1
print(happy)
```

**Explanation:** The first line, with the sizes `n` and `m`, is read into `_` and ignored. The next lines give the array `s` and two sets of integers: `A`, the liked values, and `B`, the disliked ones. The program goes through `s` once and adds 1 to `happy` for every element in `A` and subtracts 1 for every element in `B`. Sets make each membership test `i in A` fast, so the whole program runs in a single pass. A value that appears several times in `s` is counted each time, which is why `s` stays a list and only `A` and `B` are sets.

## 3. Symmetric Difference

*Difficulty: Easy*

```python
def difference():
    _ = input()
    a = set(map(int, input().split()))
    _ = input()
    b = set(map(int, input().split()))
    
    print('\n'.join(map(str, sorted(a ^ b))))
    
if __name__ == "__main__":
    difference()
```

**Explanation:** The program reads two sets of integers, each preceded by a line giving its size. The sizes are read into `_` and ignored, because the sets can be built directly from the numbers on the next line. The operator `a ^ b` computes the symmetric difference, meaning the values that belong to exactly one of the two sets. `sorted()` puts them in ascending order, and `'\n'.join(map(str, ...))` converts them to strings and prints them one per line in a single call.

## 4. Set .add()

*Difficulty: Easy*

```python
def solve():
    n = int(input())
    states = set()
    for _ in range(n):
        states.add(input())
    print(len(states))
    
if __name__ == "__main__":
    solve()
```

**Explanation:** The program reads `n`, the number of country stamps, and then reads `n` country names one per line. Each name is inserted into an initially empty `set` with `.add()`. A set keeps only one copy of each value, so repeated countries are ignored automatically. Printing `len(states)` therefore gives the number of distinct countries without any manual duplicate check.

## 5. Set .discard(), .remove() & .pop()

*Difficulty: Easy*

```python
n = int(input())
s = set(map(int, input().split()))
b = int(input())

for _ in range(b):
    text = input().split()
    command = text[0]

    if command == "pop":
        s.pop()
    elif command == "remove":
        s.remove(int(text[1]))
    elif command == "discard":
        s.discard(int(text[1]))

print(sum(s))
```

**Explanation:** The program reads the set `s`, then the number of commands `b`, and applies each command line in turn. Every line is split into words: the first word is the command name and, for `remove` and `discard`, the second is the integer to act on. `pop()` removes an arbitrary element, `remove(x)` deletes `x` and raises an error if it is missing, and `discard(x)` deletes `x` silently if it is present. After all the commands, `print(sum(s))` outputs the sum of the elements left in the set.

## 6. Set .union() Operation

*Difficulty: Easy*

```python
# Enter your code here. Read input from STDIN. Print output to STDOUT
_ = int(input())
set_a = set(map(int, input().split()))
_ = int(input())
set_b = set(map(int, input().split()))

print(len(set_a.union(set_b)))
```

**Explanation:** The program reads two groups of students, for example the subscribers to an English and to a French newspaper, each given as a size line followed by a line of roll numbers. The sizes are read into `_` and ignored, because the sets can be built directly from the numbers. `set_a.union(set_b)` returns the set of all values that appear in at least one of the two, and a student who subscribes to both is counted only once. Printing `len(...)` of that union gives the total number of distinct subscribers.

## 8. Set .difference() Operation

*Difficulty: Easy*

```python
# Enter your code here. Read input from STDIN. Print output to STDOUT
_ = int(input())
set_a = set(map(int, input().split()))
_ = int(input())
set_b = set(map(int, input().split()))

print(len(set_a.difference(set_b)))
```

**Explanation:** The program reads two groups of students, each given as a size line followed by a line of roll numbers. The sizes are read into `_` and ignored, because the sets can be built directly from the numbers. `set_a.difference(set_b)` returns the values that are in `set_a` but not in `set_b`, which here are the students subscribed only to the first newspaper. Unlike union and intersection, the order matters, so swapping the two sets would give a different answer. Printing `len(...)` of the result gives the number of those students.

## 10. Set Mutations

*Difficulty: Easy*

```python
_ = int(input())
s = set(map(int, input().split()))
b = int(input())

for _ in range(b):
    parts = input().split()
    command = parts[0]
    other_sets = set(map(int, input().split()))
    if command == "update":
        s.update(other_sets)
    elif command == "intersection_update":
        s.intersection_update(other_sets)
    elif command == "symmetric_difference_update":
        s.symmetric_difference_update(other_sets)
    elif command == "difference_update":
        s.difference_update(other_sets)

print(sum(s))
```

**Explanation:** The program reads the starting set `s`, then the number of operations `b`, and processes each operation as a pair of input lines. The first line holds the command name (its length is ignored) and the second holds the integers of the other set. Depending on the command, the matching in-place method is called: `update` adds the other set's elements, `intersection_update` keeps only the common ones, `symmetric_difference_update` keeps the values found in exactly one of the two sets, and `difference_update` removes the other set's elements. These methods modify `s` directly instead of creating a new set, and at the end `print(sum(s))` outputs the sum of the elements left.

## 11. The Captain's Room

*Difficulty: Easy*

```python
# Enter your code here. Read input from STDIN. Print output to STDOUT
k = int(input())
s = list(map(int, input().split()))

captain_room = (sum(set(s)) * k - sum(s)) // (k - 1)
print(captain_room)
```

**Explanation:** The input gives a group size `k` and a list of room numbers in which every guest room appears exactly `k` times, except the captain's room, which appears once. The solution uses arithmetic instead of counting. If `C` is the captain's room and `S` is the sum of the other distinct rooms, then `sum(set(s))` equals `S + C` and `sum(s)` equals `k·S + C`. Computing `sum(set(s)) * k - sum(s)` cancels `S` and leaves `(k - 1)·C`, so dividing by `k - 1` with `//` gives the captain's room, with no loop or dictionary needed.

## 12. Check Subset

```python
# Enter your code here. Read input from STDIN. Print output to STDOUT
k = int(input())
for x in range(k):
    _ = int(input())
    A = set(map(int, input().split()))
    _ = int(input())
    B = set(map(int, input().split()))
    print((A.issubset(B)))
```

**Explanation:** The first line gives `k`, the number of test cases, and the loop runs once per case. In each one, the program reads two sets `A` and `B`, each preceded by a size line that is read into `_` and ignored. `A.issubset(B)` returns `True` if every element of `A` is also in `B`, and `False` otherwise. The result is printed on its own line for each test case.

## 13. Check Strict Superset

*Difficulty: Easy*

```python
# Enter your code here. Read input from STDIN. Print output to STDOUT
s = set(map(int, input().split()))
k = int(input())

lista = []

for i in range(k):
    a = set(map(int, input().split()))
    lista.append(a)

check = True
for i in lista:
    check = s > i # or s.issuperset(i) if not strict
    if check is False:
        break

print(check)
```

**Explanation:** The program reads a set `s` and the number of other sets `k`, then reads those `k` sets into a list. It then goes through the list and tests, for each one, whether `s` is a strict superset using `s > i`, which is `True` only when `s` contains every element of the other set and at least one more. As soon as one test gives `False`, the loop stops with `break`, since a single failure already makes the final answer `False`. At the end `print(check)` outputs `True` only if `s` was a strict superset of every set.
