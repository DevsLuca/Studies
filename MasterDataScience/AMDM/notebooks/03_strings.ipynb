## 1. sWAP cASE

*Difficulty: Easy*

```python
def swap_case(s):
    result = []
    
    for char in s:
        if char.islower():
            result.append(char.upper())
        elif char.isupper():
            result.append(char.lower())
        else:
            result.append(char)
    
    return "".join(result)

f __name__ == '__main__':
    s = input()
    result = swap_case(s)
    print(result)
```

**Explanation:** The function `swap_case(s)` goes through the string one character at a time and collects the results in a list called `result`. Lowercase letters are converted with `.upper()`, uppercase letters with `.lower()`, and every other character (spaces, digits, punctuation) is appended unchanged. `"".join(result)` then glues the characters back into a single string. Building a list and joining once is cleaner than concatenating strings repeatedly, and Python's built-in `swapcase()` would give the same result.

## 3. Mutations

*Difficulty: Easy*

```python
def mutate_string(string, position, character):
    l = list(string)
    l[position] = character
    new_string = ''.join(l)
    return new_string

if __name__ == '__main__':
    s = input()
    i, c = input().split()
    s_new = mutate_string(s, int(i), c)
    print(s_new)
```

**Explanation:** Strings in Python are immutable, so assigning to `string[position]` would raise an error. The function first converts the string into a list with `list(string)`, which can be modified, replaces the item at the given index with the new character, and rebuilds the string with `''.join(l)`. The new string is returned and the original is left unchanged. This list round trip is the usual way to "edit" a string.

## 4. Find a string

*Difficulty: Easy*

```python
def count_substring(string, sub_string):
    count = 0
    sub_len = len(sub_string)
    for i in range(len(string) - sub_len + 1):
        if string[i : i + sub_len] == sub_string:
            count += 1
            
    return count

if __name__ == '__main__':
    string = input().strip()
    sub_string = input().strip()
    
    count = count_substring(string, sub_string)
    print(count)
```

**Explanation:** The function counts how many times `sub_string` appears in `string`, including overlapping occurrences. It slides a window of length `sub_len` along the string: the loop runs from index 0 to `len(string) - sub_len`, where the `+ 1` in the `range()` includes the last valid starting position. At each index it compares the slice `string[i : i + sub_len]` with `sub_string` and increments `count` on a match. Because the window moves one character at a time, overlaps such as `ABA` inside `ABABA` are counted, which the built-in `str.count()` would miss.

## 5. String Validators

*Difficulty: Easy*

```python
if __name__ == '__main__':
    s = input()
    print(any(c.isalnum() for c in s))
    print(any(c.isalpha() for c in s))
    print(any(c.isdigit() for c in s))
    print(any(c.islower() for c in s))
    print(any(c.isupper() for c in s))
```

**Explanation:** The program reads a string `s` and checks whether it contains at least one character of five different types: alphanumeric, alphabetic, digit, lowercase and uppercase. For each type, `any(c.method() for c in s)` goes through the characters and returns `True` as soon as one satisfies the string method (`isalnum`, `isalpha`, `isdigit`, `islower`, `isupper`), or `False` if none does. The five results are printed on separate lines in the required order. Using `any()` with a generator avoids explicit loops and stops at the first match.

**Explanation:** The program draws the HackerRank logo out of the letter `H`, using `rjust`, `ljust` and `center` to position each row. The input `thickness` controls every dimension, and the logo is built in five parts, each with its own loop: the top cone, the top pillars, the middle belt, the bottom pillars and the bottom cone. In the cones, the number of `H` characters on each side of the centre grows (or shrinks) by one per row, while the pillars and belt are centred in widths that are multiples of `thickness`. ## 7. Text Wrap

*Difficulty: Easy*

```python
def wrap(string, max_width):
    output = ""
    for i in range(0, len(string), max_width):
        output += string[i : i + max_width] + '\n'
    return output

if __name__ == '__main__':
    string, max_width = input(), int(input())
    result = wrap(string, max_width)
    print(result)
```

**Explanation:** The function `wrap(string, max_width)` cuts the string into pieces of at most `max_width` characters and puts each piece on its own line. `range(0, len(string), max_width)` steps through the string in jumps of `max_width`, and the slice `string[i : i + max_width]` extracts the piece starting at each position. A newline is appended to every piece before adding it to `output`. The last piece may be shorter, and slicing past the end of a string is safe in Python, so no special case is needed. The standard library's `textwrap` module could do this too, but the manual version shows the slicing logic.

## 8. Designer Door Mat

*Difficulty: Easy*

```python
# Enter your code here. Read input from STDIN. Print output to STDOUT
if __name__ == '__main__':
    N, M = map(int, input().split())
    
    for i in range(1, N, 2):
        print((".|." * i).center(M, "-"))

    print("WELCOME".center(M, "-"))
    
    for i in range(N - 2, -1, -2):
        print((".|." * i).center(M, "-"))
```

**Explanation:** The program prints a door mat with `N` rows and `M` columns, where each row is padded with hyphens. In the top half, a loop over the odd numbers `1, 3, ..., N-2` repeats the pattern `.|.` `i` times and centres it with `.center(M, "-")`. The middle row is the word `WELCOME` centred the same way. The bottom half mirrors the top by counting down from `N-2` to 1 in steps of 2, so the mat is symmetric around the middle row.

## 9. String Formatting

*Difficulty: Easy*

```python
def print_formatted(number):
    column = len(bin(number)[2:])
    for i in range(1, number + 1):
        d = str(i).rjust(column)
        o = oct(i)[2:].rjust(column)
        h = hex(i)[2:].upper().rjust(column)
        b = bin(i)[2:].rjust(column)
        
        print(f"{d} {o} {h} {b}")

if __name__ == '__main__':
    n = int(input())
    print_formatted(n)
```

**Explanation:** The function `print_formatted(number)` prints the numbers from 1 to `number` in four bases (decimal, octal, hexadecimal and binary) as aligned columns. The width of every column is the length of the binary representation of the largest number, `len(bin(number)[2:])`, where `[2:]` removes the `0b` prefix. For each `i`, the code converts it with `oct`, `hex` (uppercased) and `bin`, strips the prefixes, and right-aligns each value with `rjust(column)`. The four values are then printed on one line, separated by spaces.

## 10. Alphabet Rangoli

*Difficulty: Easy*

```python
def print_rangoli(size):
    width = (size*2-1)+(size*2-2)
    rows = []
    
    for i in range(size):
        row_chars = []
        for j in range(i + 1):
            char_code = ord('a') + size - 1 - j
            row_chars.append(chr(char_code))
            
        for j in range(i):
            char_code = ord('a') + size - i + j
            row_chars.append(chr(char_code))
            
        row_str = "-".join(row_chars)
        rows.append(row_str.center(width, "-"))
        
    full_rangoli = rows + rows[:-1][::-1]
    print("\n".join(full_rangoli))
    
if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
```

**Explanation:** The function `print_rangoli(size)` builds a symmetric pattern of letters joined by hyphens, with the largest letter at the edges and `a` in the centre. Each row is built from a descending run of letters starting at the largest one (the first inner loop) followed by an ascending run (the second), joined with `"-".join(...)` and centred in a line of `width` hyphens. The top half and centre row are stored in `rows`, and the bottom half is the same list without its last element, reversed (`rows[:-1][::-1]`). The rows are joined with newlines and printed in one call.

## 12. The Minion Game

*Difficulty: Medium*

```python
def minion_game(string):
    vocali = "AEIOU"
    kevin_points = 0
    stuart_points = 0
    
    for i in range(len(string)):
        if string[i] in vocali:
            kevin_points += (len(string)) - i
        else:
            stuart_points += (len(string)) - i
    
    if stuart_points < kevin_points:
        print(f"Kevin {kevin_points}")
    elif stuart_points > kevin_points:
        print(f"Stuart {stuart_points}")
    else:
        print("Draw")

if __name__ == '__main__':
    s = input()
    minion_game(s)
```

**Explanation:** Kevin scores with substrings that start with a vowel and Stuart with those that start with a consonant, and every occurrence of a substring counts as a point. Instead of generating all substrings, the code uses the fact that a letter at index `i` is the start of exactly `len(string) - i` substrings. For each position it adds that number to Kevin's score if the letter is in `"AEIOU"`, and to Stuart's score otherwise. At the end it compares the two scores and prints the winner with their points, or `Draw`. This runs in one pass over the string instead of building every substring.

## 13. Merge the Tools!

*Difficulty: Medium*

```python
def merge_the_tools(string, k):
    lista = []

    for i in range(0, len(string), k):
        chunks = string[i:i+k]
        lista.append(chunks)

    for x in range(len(lista)):
        seen = ""
        for char in lista[x]:
            if char in seen:
                continue
            else:
                seen += char
        print(seen)

if __name__ == '__main__':
    string, k = input(), int(input())
    merge_the_tools(string, k)
```

**Explanation:** The function first cuts the string into consecutive chunks of length `k`, using `range(0, len(string), k)` and slicing, and stores them in the list `lista`. For each chunk, it builds a new string `seen` by going through the characters in order and adding a character only if it is not already in `seen`. This removes duplicates while keeping the order of first appearance. Each resulting string is printed on its own line.

## 14. String Split and Join

*Difficulty: Easy*

```python
def split_and_join(line):
    line = line.split(" ")
    line = "-".join(line)
    return line
    
if __name__ == '__main__':
    line = input()
    result = split_and_join(line)
    print(result)
```

**Explanation:** The function `split_and_join(line)` first breaks the input into a list of words with `line.split(" ")`, cutting at every space. It then uses `"-".join(line)` to glue the words back together with a hyphen as the separator. Since strings are immutable, each step produces a new object, and the variable `line` is simply reused to hold it. Splitting on one delimiter and joining with another is the standard way to replace the separators in a string.
