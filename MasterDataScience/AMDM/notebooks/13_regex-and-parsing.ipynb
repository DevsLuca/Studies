## 1. Detect Floating Point Number

*Difficulty: Easy*

```python
import re

for _ in range(int(input())):
    print(bool(re.match(r"^[+-]?\d*\.\d+$", input())))
```

**Explanation:** The program reads the number of test cases and, for each string, prints `True` if it is a valid floating-point number and `False` otherwise. The regular expression `^[+-]?\d*\.\d+$` is built from four parts: an optional sign (`[+-]?`), any number of digits before the decimal point (`\d*`), a literal dot (`\.`), and at least one digit after it (`\d+`). The mandatory dot and the required digit after it reject plain integers such as `12` and numbers that end with a dot such as `4.`. `re.match` returns a match object or `None`, and `bool()` turns that into the `True` or `False` that is printed.

## 2. Re.split()

*Difficulty: Easy*

```python
regex_pattern = r"[.,]"	# Do not delete 'r'.

import re

print("\n".join(re.split(regex_pattern, input())))
```

**Explanation:** The task is to split a string at every comma and every dot. The pattern `[.,]` is a character class that matches either one of the two characters, and inside the brackets the dot is a literal dot, so it does not need escaping. `re.split(regex_pattern, input())` cuts the input at each match and returns the pieces as a list, and `"\n".join(...)` puts each piece on its own line. The `r` in front of the pattern string makes it a raw string, which is the habit to follow for regular expressions so backslashes reach the regex engine unchanged.

## 3. Group(), Groups() & Groupdict()

*Difficulty: Easy*

```python
import re

s = input()

check = re.search(r"([a-zA-Z0-9])\1+", s)

if check:
    print(check.group(1))
else:
    print(-1)
```

**Explanation:** The program finds the first alphanumeric character that appears twice or more in a row in the input string, and prints it, or `-1` if there is none. The pattern `([a-zA-Z0-9])\1+` has two parts. The first captures one letter or digit in a group, and the second, `\1+`, is a backreference that matches the same character again, one or more times. `re.search` returns the leftmost match, so the first repeated character in the string is the one found. `check.group(1)` then gives the character captured by the first group, and if there is no match, `check` is `None` and `-1` is printed.

## 4. Re.findall() & Re.finditer()

*Difficulty: Easy*

```python
import re

s = input()

pattern = r"(?<=[qwrtypsdfghjklzxcvbnm])[aeiou]{2,}(?=[qwrtypsdfghjklzxcvbnm])" 

check = list(re.finditer(
        pattern,
        s,
        re.IGNORECASE))

if check:
    for i in check:
        print(i.group(0))
else:
    print(-1)
```

**Explanation:** The task is to find every substring of two or more vowels that sits between two consonants, and print each one on its own line, or `-1` if there are none. The pattern `(?<=[consonants])[aeiou]{2,}(?=[consonants])` has three parts: a lookbehind that requires a consonant just before, the run of at least two vowels, and a lookahead that requires a consonant just after. Lookarounds only check the neighbouring characters without consuming them, so the consonants are not part of the match and can be shared by two adjacent runs. `re.finditer` returns the matches in order, `re.IGNORECASE` makes the search work for capital letters, and `i.group(0)` prints the text of each match.

## 5. Re.start() & Re.end()

*Difficulty: Easy*

```python
import re

a = input()
b = input()

check = list(re.finditer(f"(?=({b}))", a))

if check:
    for i in check:
        print((i.start(0), i.end()+len(b)-1))
else:
    print((-1, -1))
```

**Explanation:** The task is to find every occurrence of the substring `b` in the string `a`, including overlapping ones, and print the start and end index of each as a tuple, or `(-1, -1)` if there is none. The pattern `(?=(b))` is a lookahead, which matches at every position where `b` begins without consuming any characters, so overlapping occurrences are not skipped. Because each match has zero width, `i.start(0)` is the index where the occurrence begins, and `i.end() + len(b) - 1` gives the index of its last character. The tuples are printed in order of appearance.

## 6. Regex Substitution

*Difficulty: Medium*

```python
import re

n = int(input())

for _ in range(n):
    line = input()
    line = re.sub(r"(?<=\s)&&(?=\s)", "and", line)
    line = re.sub(r"(?<=\s)\|\|(?=\s)", "or", line)
    print(line)
```

**Explanation:** The task is to replace `&&` with `and` and `||` with `or` in each line of text, but only when the operator has a space on both sides. The pattern `(?<=\s)&&(?=\s)` uses a lookbehind and a lookahead to require whitespace on each side without including it in the match, so the surrounding spaces stay untouched and operators glued to other characters, such as `a&&b`, are left alone. In the second pattern the pipes must be escaped as `\|\|`, because `|` means "or" in regular expressions. Each line goes through two `re.sub` calls, one per operator, and is then printed.

## 7. Validating Roman Numerals

*Difficulty: Easy*

```python
regex_pattern = r"^M{0,3}(C[MD]|D?C{0,3})(X[CL]|L?X{0,3})(I[VX]|V?I{0,3})$"
# Do not delete 'r'.

import re
print(str(bool(re.match(regex_pattern, input()))))
```

**Explanation:** The pattern checks a Roman numeral digit by digit, from thousands down to units, and the anchors `^` and `$` force it to match the whole string. `M{0,3}` allows up to three thousands. The hundreds, tens and ones each use the same shape: a special subtractive case (`C[MD]` for 900 and 400, `X[CL]` for 90 and 40, `I[VX]` for 9 and 4), or an optional five-symbol (`D?`, `L?`, `V?`) followed by up to three repeats of the basic symbol (`C{0,3}`, `X{0,3}`, `I{0,3}`). Since each alternative covers exactly the legal forms for that position, invalid strings such as `IIII` or `VX` are rejected. `bool(re.match(...))` turns the result into `True` or `False`, and `str()` makes it printable text.

## 8. Validating phone numbers

*Difficulty: Easy*

```python
import re

n = int(input())

for _ in range(n):
    phone = input()
    if re.match(r"^[789]\d{9}$", phone):
        print("YES")
    else:
        print("NO")
```

**Explanation:** The task is to decide whether each input string is a valid mobile number, meaning exactly ten digits that start with 7, 8 or 9. The pattern `^[789]\d{9}$` expresses this directly: `[789]` accepts one of the three allowed first digits, `\d{9}` requires exactly nine more digits, and the anchors `^` and `$` force the whole string to match, so numbers that are too short, too long, or that contain other characters are rejected. For each of the `n` strings, the program prints `YES` if `re.match` finds a match and `NO` otherwise.

## 9. Validating and Parsing Email Addresses

*Difficulty: Easy*

```python
import re
import email.utils

pattern = r"^[a-zA-Z][\w\.-]*@[a-zA-Z]+\.[a-zA-Z]{1,3}$"

n = int(input())

for _ in range(n):
    data = input()
    name, email_address = email.utils.parseaddr(data)

    if re.match(pattern, email_address):
        print(email.utils.formataddr((name, email_address)))
```

**Explanation:** Each input line has the form `name <email>`, and the program prints it back only when the email address is valid. The standard library function `email.utils.parseaddr` splits the line into the display name and the address, so no manual string cutting is needed. The pattern `^[a-zA-Z][\w\.-]*@[a-zA-Z]+\.[a-zA-Z]{1,3}$` checks the rules in order: a username that starts with a letter and continues with letters, digits, underscores, dots or hyphens, then `@`, a domain made only of letters, a dot, and an extension of one to three letters. For valid addresses, `email.utils.formataddr` rebuilds the original `name <email>` format, and invalid lines produce no output.

## 10. Hex Color Code

*Difficulty: Easy*

```python
import re

n = int(input())
flag = False

for _ in range(n):
    line = input()
    if "{" in line:
        flag = True
    if "}" in line:
        flag = False
    
    if flag and "{" not in line:
        check = re.findall(r"#[\da-fA-F]{3,6}", line)
        for i in check:
            print(i)
```

**Explanation:** The task is to print every valid hex color code that appears inside the CSS declaration blocks, in the order they occur. The variable `flag` records whether the current line is inside a block: it turns on when a line contains `{` and off when a line contains `}`. The first line of a block, which holds the selector, is skipped so that selectors like `#BED` are not mistaken for colors. On the lines inside a block, `re.findall(r"#[\da-fA-F]{3,6}", line)` returns every `#` followed by three to six hexadecimal digits, and each result is printed on its own line.

## 11. HTML Parser - Part 1

*Difficulty: Easy*

```python
from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):

  def handle_starttag(self, tag, attrs):
    print("Start :", tag)
    for attr, val in attrs:
      print("->", attr, ">", val)

  def handle_endtag(self, tag):
    print("End   :", tag)  # Two extra spaces here for alignment

  def handle_startendtag(self, tag, attrs):
    print("Empty :", tag)
    for attr, val in attrs:
      print("->", attr, ">", val)


parser = MyHTMLParser()
parser.feed("".join([input().strip() for _ in range(int(input()))]))
```

**Explanation:** The program defines a subclass of Python's `HTMLParser` and overrides three of its methods, so the parser calls them automatically each time it meets a start tag, an end tag or an empty tag. In every method it prints a label (`Start :`, `End   :` or `Empty :`) followed by the tag name, and for tags with attributes it loops over the `(name, value)` pairs in `attrs` and prints each one in the form `-> name > value`. The main block reads the `n` lines of HTML, removes the whitespace around each one with `.strip()`, joins them into a single string, and passes it to `parser.feed()`, which runs the parser over the whole text. An attribute without a value is reported with the value `None`, which is what the exercise expects.

## 12. HTML Parser - Part 2

*Difficulty: Easy*

```python
from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):

  def handle_comment(self, data):
    if "\n" in data:
      print(">>> Multi-line Comment")
    else:
      print(">>> Single-line Comment")
    print(data)

  def handle_data(self, data):
    if data.strip():
      print(">>> Data")
      print(data)


html = ""
for i in range(int(input())):
  html += input().rstrip() + "\n"

parser = MyHTMLParser()
parser.feed(html)
parser.close()
```

**Explanation:** The program subclasses `HTMLParser` and overrides two methods: `handle_comment`, which is called for every comment, and `handle_data`, which is called for every piece of text between tags. A comment is multi-line if it contains a newline character, so the code prints `>>> Multi-line Comment` or `>>> Single-line Comment` accordingly and then prints the comment text. For data, the line `if data.strip():` skips pieces that contain only whitespace, such as the line breaks between tags, and prints `>>> Data` followed by the text only for real content. The main block reads the `n` lines, keeps a newline after each one so multi-line comments stay detectable, and passes the whole text to `parser.feed()`, with `parser.close()` finishing the parse.
