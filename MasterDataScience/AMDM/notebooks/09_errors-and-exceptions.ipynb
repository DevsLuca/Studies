## 1. Exceptions

*Difficulty: Easy*

```python
n = int(input())
for _ in range(n):
    a, b, = input().split()
    try:
        fine = int(a)//int(b)
        print(fine)
    except ZeroDivisionError:
        print("Error Code: integer division or modulo by zero") 
        # Hackerrank has an older automated message
    except ValueError as e:
        print("Error Code:", e)
```

**Explanation:** The program reads `n` test cases, each with two values `a` and `b` that arrive as strings. Inside a `try` block it converts them with `int()` and prints the result of the integer division `int(a)//int(b)`. If `b` is zero, the `ZeroDivisionError` handler prints the fixed message `Error Code: integer division or modulo by zero`. If either value is not a valid integer, `int()` raises a `ValueError`, and the second handler prints `Error Code:` followed by the error's own message, which it receives through `as e`. Each test case produces exactly one line of output, either the quotient or one of the two error messages.

## 2. Incorrect Regex

*Difficulty: Easy*

**What the challenge asks.** For each input string, print `True` if it is a valid regular expression and `False` if not. The sample input uses `.*+` as the invalid example, because it fails with a "multiple repeat" error: a quantifier (`*`) is followed directly by another one (`+`).

**What changed between Python versions.** The challenge was written when `re.compile(".*+")` raised an error. Possessive quantifiers were added to Python's `re` module in version 3.11, and they are written as a quantifier followed by `+`, such as `*+`, `++`, `?+` and `{m,n}+`. The same pattern therefore behaves differently depending on the version:

| Pattern | Python up to 3.10 | Python 3.11 and later |
|---|---|---|
| `.*+` | `re.error`, so `False` | valid (possessive), so `True` |
| `.++` | `re.error`, so `False` | valid (possessive), so `True` |

The approach (`re.compile()` inside a `try/except re.error` block) is the same in every version, but the output for patterns like `.*+` depends on the interpreter.
