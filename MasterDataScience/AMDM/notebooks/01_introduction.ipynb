## 2. Python If-Else

*Difficulty: Easy*

```python
import math
import os
import random
import re
import sys

if __name__ == '__main__':
    n = int(input().strip())
    
    if n % 2 != 0:
        print("Weird")
    else:
        if 2 <= n <= 5:
            print("Not Weird")
        elif 6 <= n <= 20:
            print("Weird")
        elif n > 20:
            print("Not Weird")
```

**Explanation:** The program reads an integer `n` and prints `Weird` or `Not Weird` depending on its parity and range. The modulo operator (`n % 2 != 0`) detects odd numbers, which are always `Weird`. For even numbers, an `if/elif` chain checks the ranges: 2 to 5 is `Not Weird`, 6 to 20 is `Weird`, and anything above 20 is `Not Weird`. The `.strip()` call removes stray whitespace from the input line before it is converted to an integer.

## 3. Arithmetic Operators

*Difficulty: Easy*

```python
if __name__ == '__main__':
    a = int(input())
    b = int(input())
    c = a + b
    d = a - b
    e = a * b
    print(c)
    print(d)
    print(e)
```

**Explanation:** The program reads two integers, `a` and `b`, converting each with `int()` because `input()` always returns a string. It then computes their sum, difference and product with the operators `+`, `-` and `*`, storing the results in `c`, `d` and `e`. Finally, it prints the three results on separate lines in the order the problem requires.

## 4. Python: Division

*Difficulty: Easy*

```python
if __name__ == '__main__':
    a = int(input())
    b = int(input())
    print(a//b)
    print(a/b)
```

**Explanation:** The program reads two integers and prints two results: first the integer (floor) division `a // b`, then the float division `a / b`. The `//` operator discards the fractional part, while `/` always returns a floating-point number. Printing both with the same operands makes the difference between the two operators easy to see.

## 5. Loops

*Difficulty: Easy*

```python
if __name__ == '__main__':
    n = int(input())
    i = 0
    while i < n:
        print(i*i)
        i += 1
```

**Explanation:** The program reads an integer `n` and prints the square of every integer from 0 up to `n - 1`, one per line. It uses a `while` loop with a counter `i` that starts at 0 and runs as long as `i < n`. Each pass prints `i * i` and then increments `i` by 1. Without the increment, the condition would never become false and the loop would run forever.

## 6. Write a function

*Difficulty: Medium*

```python
def is_leap(year):
    leap = False
    
    if year % 4 == 0:
        if year % 100 == 0:
            if year % 400 == 0:
                leap = True
        else:
            leap = True
            
    return leap

year = int(input())
print(is_leap(year))
```

**Explanation:** The function `is_leap(year)` returns `True` if a year is a leap year and `False` otherwise, following the Gregorian rules. A year divisible by 4 is a candidate, but a century year (divisible by 100) is only a leap year if it is also divisible by 400. A flag `leap` starts as `False` and is switched to `True` only in the cases that qualify. For example, 2000 is a leap year, while 1900 is not.

## 7. Print Function

*Difficulty: Easy*

```python
if __name__ == '__main__':
    n = int(input())
    print(*(range(1, n + 1)), sep='')
```

**Explanation:** The program reads an integer `n` and prints the numbers from 1 to `n` joined together on a single line with no spaces. `range(1, n + 1)` generates the numbers 1 to `n`, and the `+ 1` is needed because the end of a range is exclusive. The `*` operator unpacks the range into separate arguments for `print()`, and `sep=''` removes the default space between them. This avoids building the string with a loop.
