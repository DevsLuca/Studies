## 1. Map and Lambda Function

*Difficulty: Easy*

```python
cube = lambda x: x**3

def fibonacci(n):
    lista = []
    for i in range(0, n):
        if i == 0:
            lista.append(i)
        if i == 1:
            lista.append(i)
        if i > 1:
            out = lista[i-1] + lista[i-2]
            lista.append(out)
    return lista
    # return a list of fibonacci numbers

if __name__ == '__main__':
    n = int(input())
    print(list(map(cube, fibonacci(n))))
```

**Explanation:** The program defines `cube` as a lambda, an anonymous one-line function that returns `x**3`. The function `fibonacci(n)` builds the first `n` Fibonacci numbers in a list: the first two values are 0 and 1, and every later value is the sum of the two before it (`lista[i-1] + lista[i-2]`). In the main block, `map(cube, fibonacci(n))` applies the lambda to every element of that list without writing a loop, and `list(...)` turns the result into a list so it can be printed. The output is the list of the cubes of the first `n` Fibonacci numbers.
