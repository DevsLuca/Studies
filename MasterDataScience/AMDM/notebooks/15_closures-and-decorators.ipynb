## 1. Standardize Mobile Number Using Decorators

*Difficulty: Easy*

```python
def wrapper(f):
    def fun(l):
        # complete the function
        lista = []
        for num in l:
            lista.append("+91 " + num[-10:-5] + " " + num[-5:])
        return f(lista)
    return fun

@wrapper
def sort_phone(l):
    print(*sorted(l), sep='\n')

if __name__ == '__main__':
    l = [input() for _ in range(int(input()))]
    sort_phone(l)
```

**Explanation:** The function `wrapper(f)` is a decorator: it takes a function `f` and returns a new function `fun` that first cleans up the list of phone numbers and then passes the result to `f`. For each number, it keeps only the last ten digits with `num[-10:]`, whatever prefix the number had (`+91`, `91`, `0` or none), and rebuilds it as `+91 xxxxx xxxxx` by splitting those digits into two groups of five. The `@wrapper` line above `sort_phone` is shorthand for `sort_phone = wrapper(sort_phone)`, so calling `sort_phone(l)` standardizes the numbers automatically. The original `sort_phone` then prints them in sorted order, one per line.

## 2. Decorators 2 - Name Directory

*Difficulty: Easy*

```python
import operator
from operator import itemgetter

def person_lister(f):
    def inner(people):
        sorted_people = sorted(people, key=lambda x: int(x[2]))
        formatted_people = []
        for person in sorted_people:
            formatted_people.append(f(person))
        return formatted_people
    return inner

@person_lister
def name_format(person):
    return ("Mr. " if person[3] == "M" else "Ms. ") + person[0] + " " + person[1]

if __name__ == '__main__':
    people = [input().split() for i in range(int(input()))]
    print(*name_format(people), sep='\n')
```

**Explanation:** The decorator `person_lister(f)` returns a function `inner` that takes the whole list of people, sorts it by age, and then applies `f` to each person. The sort uses `key=lambda x: int(x[2])`, which converts the age field to an integer so ages compare as numbers and not as text. Python's sort is stable, so people with the same age stay in the order they were entered, as the exercise requires. The decorated function `name_format` builds the display name from one person's record, adding `Mr.` or `Ms.` depending on the gender field, followed by first and last name, and the results are printed one per line.
