## 1. Birthday Cake Candles

*Difficulty: Easy* · [HackerRank challenge](https://www.hackerrank.com/challenges/birthday-cake-candles)

```python
import math
import os
import random
import re
import sys

def birthdayCakeCandles(candles):
    tallest_candle = max(candles)
    count = 0
    for i in candles:
        if i == tallest_candle:
            count += 1            
    return count


if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')
    candles_count = int(input().strip())
    candles = list(map(int, input().rstrip().split()))
    result = birthdayCakeCandles(candles)
    fptr.write(str(result) + '\n')
    fptr.close()
```

**Explanation:** The function receives the heights of the candles on a cake and must return how many of them are as tall as the tallest one, since only those can be blown out. It first finds the maximum height with `max(candles)`, then goes through the list once and increases `count` every time a candle has exactly that height. The count is returned at the end, so ties for the tallest candle are all included. The input line with the number of candles is read but not needed, because the list already carries its own length.

**Complexity:** O(n) time, since the list is traversed twice (once by `max` and once by the loop), and O(1) extra space.

## 2. Number Line Jumps (kangaroo)

*Difficulty: Easy* · [HackerRank challenge](https://www.hackerrank.com/challenges/kangaroo)

```python
def kangaroo(x1, v1, x2, v2):
    if v2 >= v1:
        result = "NO"
    else:
        if (x2 - x1)%(v2 - v1) == 0:
            result = "YES"
        else:
            result = "NO"
    return result

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')
    first_multiple_input = input().rstrip().split()
    x1 = int(first_multiple_input[0])
    v1 = int(first_multiple_input[1])
    x2 = int(first_multiple_input[2])
    v2 = int(first_multiple_input[3])
    result = kangaroo(x1, v1, x2, v2)
    fptr.write(result + '\n')
    fptr.close()
```

**Explanation:** The two kangaroos start at `x1` and `x2` (with `x1 < x2`) and jump `v1` and `v2` metres per jump, so they meet after `t` jumps when `x1 + v1*t = x2 + v2*t`. If `v2 >= v1`, the kangaroo in front is at least as fast as the one behind, so the gap never closes and the answer is `"NO"`. Otherwise the gap `x2 - x1` shrinks by `v1 - v2` with every jump, so they land on the same spot exactly when the gap is divisible by the difference in speeds, which is the test `(x2 - x1) % (v2 - v1) == 0`. The divisor is negative here because of the order of the subtraction, but divisibility doesn't depend on the sign, so the check works.

**Complexity:** O(1) time and O(1) space, since the answer comes from a single arithmetic check and no simulation of the jumps is needed.

## 3. Viral Advertising

*Difficulty: Easy* · [HackerRank challenge](https://www.hackerrank.com/challenges/strange-advertising)

```python
import math
import os
import random
import re
import sys

def viralAdvertising(n):
    shared = 5
    cumulative = 0
    
    for _ in range(n):
        liked = shared // 2
        cumulative += liked
        shared = liked * 3
    
    return cumulative

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')
    n = int(input().strip())
    result = viralAdvertising(n)
    fptr.write(str(result) + '\n')
    fptr.close()
```

**Explanation:** On day 1 an advertisement is shown to 5 people, and every day half of the people who see it (rounded down) like it and each of them shares it with 3 friends, who see it the next day. The function simulates this process: on each of the `n` days, `liked = shared // 2` is the number of new likes, which is added to the running total `cumulative`, and `shared = liked * 3` is the audience for the following day. The integer division `//` takes care of the rounding down. After `n` days, `cumulative` holds the total number of people who liked the advertisement and is returned.

**Complexity:** O(n) time, because the loop runs once per day, and O(1) extra space, because only two integers are kept.

## 4. Recursive Digit Sum

*Difficulty: Medium* · [HackerRank challenge](https://www.hackerrank.com/challenges/recursive-digit-sum)

```python
import math
import os
import random
import re
import sys

def superDigit(n, k):
    total = 0
    for char in n:
        total += int(char)
    total *= k
    total = str(total)
    while len(total) > 1:
        return superDigit(total, 1)
    else:
        return int(total)
            
if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')
    first_multiple_input = input().rstrip().split()
    n = first_multiple_input[0]
    k = int(first_multiple_input[1])
    result = superDigit(n, k)
    fptr.write(str(result) + '\n')
    fptr.close()
```

**Explanation:** The task is to find the super digit of a number `n` written `k` times in a row, which is the single digit you get by repeatedly summing the digits. Building that huge concatenated number would be far too large, so the code uses the fact that its digit sum equals `k` times the digit sum of `n`. It sums the digits of `n`, multiplies the result by `k`, and converts it to a string. If that string has more than one digit, the function calls itself on it with `k = 1`, which repeats the digit sum until a single digit remains, and that digit is returned as an integer. The input `n` is read as a string because it can have up to 100,000 digits.

**Complexity:** O(len(n)) time, because only the first digit sum touches the long input, and every later step works on a number of about eleven digits at most. The extra space is O(1), and the recursion is only a few calls deep.

## 5. Insertion Sort - Part 1

*Difficulty: Easy* · [HackerRank challenge](https://www.hackerrank.com/challenges/insertionsort1)

```python
import math
import os
import random
import re
import sys

def insertionSort1(n, arr):
    val = arr[-1]
    i = n - 2
    
    while i >= 0 and arr[i] > val:
        arr[i + 1] = arr[i]
        print(*(arr))
        i -= 1

    arr[i + 1] = val
    print(*(arr))

if __name__ == '__main__':
    n = int(input().strip())
    arr = list(map(int, input().rstrip().split()))
    insertionSort1(n, arr)
```

**Explanation:** The array is sorted except for its last element, and the task is to insert that element into its correct place, printing the array after every shift. The code first saves the last value in `val`, because the first shift will overwrite its position, and starts a pointer `i` at the second-to-last index. While `i` is still inside the array and `arr[i]` is larger than `val`, the `while` loop copies that element one place to the right, prints the array, and moves `i` one step left. When the loop stops, `i + 1` is the gap where `val` belongs, so it is written there and the final sorted array is printed.

**Complexity:** the shifting itself is O(n) time and O(1) extra space, because each element moves at most once. Since the array is printed after every shift, the total work including output is O(n²) in the worst case.

## 6. Insertion Sort - Part 2

*Difficulty: Easy* · [HackerRank challenge](https://www.hackerrank.com/challenges/insertionsort2)

```python
import math
import os
import random
import re
import sys

def insertionSort2(n, arr):
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
        print(*(arr))

if __name__ == '__main__':
    n = int(input().strip())
    arr = list(map(int, input().rstrip().split()))
    insertionSort2(n, arr)
```

**Explanation:** The function sorts the whole array by repeating the idea of Part 1 for every element, printing the array once after each insertion. The outer loop starts at index 1, since the first element on its own is already sorted, and treats `arr[i]` as the `key` to insert into the sorted part on its left. The inner `while` loop moves backwards from `i - 1`, shifting every element larger than `key` one place to the right, and stops at the first element that is not larger or at the start of the array. The key is then written into the gap at `j + 1`, and the array is printed, so each line shows the sorted prefix growing by one element.

**Complexity:** O(n²) time in the worst case, when the array is in reverse order and every element must shift all the way to the front. On an already sorted array the sorting itself takes only O(n) comparisons, but printing the array on every pass makes the total O(n²) either way. It uses O(1) extra space, because the sort happens in place.
