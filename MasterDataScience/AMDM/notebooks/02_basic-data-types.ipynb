## 1. Find the Runner-Up Score!

*Difficulty: Easy*

```python
if __name__ == '__main__':
    n = int(input())
    arr = map(int, input().split())
    set_scores = sorted(list(set(arr)))
    print(set_scores[-2])
```

**Explanation:** The program reads `n`, the number of scores, and then the scores themselves as a list of integers. Wrapping the list in `set()` removes duplicates, so a tied top score is only counted once. `sorted()` puts the unique scores in ascending order, and `[-2]` picks the second-to-last element, which is the runner-up. Removing duplicates first is the key step, because without it the second-highest value could simply be a repeat of the winner's score.

## 2. Nested Lists

*Difficulty: Easy*

```python
if __name__ == '__main__':
    students = []
    for _ in range(int(input())):
        name = input()
        score = float(input())
        students.append([name, score])

    grades = sorted(list(set(student[1] for student in students)))
    second_lowest = grades[1]
    
    for name, score in sorted(students):
        if score == second_lowest:
            print(name)
```

**Explanation:** The program reads `n` students, each given as a name on one line and a grade on the next, converting the grade to a float. The goal is to print the names of the students with the second-lowest grade, in alphabetical order. This is done by storing each `[name, score]` pair in a nested list, finding the second-lowest unique score, and collecting every name that matches it. Sorting those names alphabetically before printing handles ties.

## 3. Finding the percentage

*Difficulty: Easy*

```python
if __name__ == '__main__':
    n = int(input())
    student_marks = {}
    for _ in range(n):
        name, *line = input().split()
        scores = list(map(float, line))
        student_marks[name] = scores
    query_name = input()
    
    query_scores = student_marks[query_name]
    average = sum(query_scores) / len(query_scores)
    print(f"{average:.2f}")
```

**Explanation:** The program reads `n` lines, each with a student's name followed by their marks, and stores them in a dictionary mapping each name to a list of floats. The `name, *line = input().split()` syntax puts the first word in `name` and all remaining words in `line`. It then reads the name to query, looks up that student's marks, and computes the average with `sum() / len()`. The `:.2f` format prints the result rounded to two decimal places.

## 4. Lists

*Difficulty: Easy*

```python
if __name__ == '__main__':
    N = int(input())
    lst = []
    
    for _ in range(N):
        command = input().split()
        cmd = command[0]

        if cmd == "insert":
            index = int(command[1])
            element = int(command[2])
            lst.insert(index, element)
            
        elif cmd == "print":
            print(lst)
            
        elif cmd == "remove":
            element = int(command[1])
            lst.remove(element)
            
        elif cmd == "append":
            element = int(command[1])
            lst.append(element)
            
        elif cmd == "sort":
            lst.sort()
            
        elif cmd == "pop":
            lst.pop()
            
        elif cmd == "reverse":
            lst.reverse()
```

**Explanation:** The program reads `N`, the number of commands, and applies each one to a list `lst` that starts empty. Each command line is split into words: the first word is the command name and the rest are its arguments, converted to integers where needed. An `if/elif` chain calls the matching list method: `insert`, `remove`, `append`, `sort`, `pop` or `reverse`, and `print` shows the current list. This pattern, parsing a command and dispatching it to the right operation, is a common way to simulate a small interpreter.

## 5. Tuples

*Difficulty: Easy*

```python
if __name__ == '__main__':
    n = int(input())
    integer_list = map(int, input().split())
    
    t = tuple(integer_list)
    print(hash(t))
```

**Explanation:** The program reads `n` and then a line of integers, which it converts to a list with `map(int, ...)`. It turns the list into a tuple with `tuple()` and prints its hash with `hash()`. Tuples are immutable, so they are hashable and can be used as dictionary keys, while lists are not. The printed value is an integer that Python derives from the tuple's contents.

## 6. List Comprehensions

*Difficulty: Easy*

```python
if __name__ == '__main__':
    x = int(input())
    y = int(input())
    z = int(input())
    n = int(input())
    coordinates = [
        [i, j, k] 
        for i in range(x+1)
        for j in range(y+1)
        for k in range(z+1)
        if i+j+k != n
        ]
    print(coordinates)
```

**Explanation:** The program reads four integers `x`, `y`, `z` and `n`, and builds the list of all coordinates `[i, j, k]` on a 3D grid, where `i` runs from 0 to `x`, `j` from 0 to `y` and `k` from 0 to `z`. A single list comprehension does the work, with three nested `for` clauses and a trailing `if i + j + k != n` that keeps only the coordinates whose sum is different from `n`. The `+ 1` in each `range()` is needed because the end of a range is exclusive, so the upper bound itself is included. The result is printed in the order the loops generate it, with `i` changing slowest and `k` fastest.
