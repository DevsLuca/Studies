## 0. Fibonacci Number (extra warm-up)

**Difficulty:** Easy | **LeetCode:** [LC0509](https://leetcode.com/problems/fibonacci-number) | [back to top](#problem-list)

**Status:** ✅

**📌 Named sequence: Fibonacci.** Each term is the sum of the two before it (0, 1, 1, 2, 3, 5, 8, ...), with `F0 = 0` and `F1 = 1`.

**Problem**

Given `n`, return the n-th Fibonacci number, where `F(n) = F(n-1) + F(n-2)`.

**Approach**

- The simplest DP: `F(n) = F(n-1) + F(n-2)`, with base cases `F0 = 0` and `F1 = 1`.
- Plain recursion is exponential because of overlapping subproblems, since `F(n-2)` is computed again inside `F(n-1)`. Building from the bottom up with two variables fixes that.
- This is the same recurrence as Climbing Stairs, shifted by one: ways(n) = F(n+1).

**Complexity:** time O(n) · space O(1)

```python
class Solution:
    def fib(self, n: int) -> int:
        if n <= 1:
            return n
        prev, curr = 0, 1
        for _ in range(2, n + 1):
            prev, curr = curr, prev + curr
        return curr
```

## 1. Climbing Stairs

**Difficulty:** Easy | **LeetCode:** [LC0070](https://leetcode.com/problems/climbing-stairs) | [back to top](#problem-list)

**Status:** ✅

**📌 Fibonacci.** Each term is the sum of the two before it (1, 2, 3, 5, 8, ...). The number of ways to climb n steps is the (n+1)-th Fibonacci number.

**Problem**

You are climbing a staircase with `n` steps. Each move you climb either 1 or 2 steps. Return the number of distinct sequences of moves that reach the top.

**Approach**

Key observation: the last move onto step `n` came either from step `n-1` (a 1-step) or from step `n-2` (a 2-step). These cases never overlap, so `ways(n) = ways(n-1) + ways(n-2)`, with base cases `ways(1) = 1` and `ways(2) = 2`.

**Solution 1: plain recursion (correct, but fails the time limit)**

- Implements the recurrence directly.
- It recomputes the same subproblems again and again, so the call tree doubles at every level.
- Time O(2^n) (more tightly O(1.618^n)) · space O(n) for the call stack.

**Solution 2: bottom-up DP with two variables (accepted)**

- Build up from step 3 to `n`, keeping only the last two values, since each step depends only on those.
- Theory: this is 1-D DP with space optimization. The full table `dp[i]` is compressed to two variables.
- Time O(n) · space O(1).

**Takeaway:** the recursion is correct but exponential because of **overlapping subproblems**. Fixing that is the whole point of DP. Two standard fixes are memoization (top-down: cache each result, O(n) time and O(n) space) and tabulation (bottom-up, which is Solution 2).

```python
class Solution1:
    def climbStairs(self, n: int) -> int:
        # 1. Base cases: If we are at step 1 or 2, we know the answer immediately
        if n == 1:
            return 1
        if n == 2:
            return 2
        
        # 2. Recursive step: The sum of the previous two steps
        return self.climbStairs(n - 1) + self.climbStairs(n - 2)
```

```python
class Solution2:
    def climbStairs(self, n: int) -> int:
        if n <= 2:
            return n
            
        # We only need to remember the last two steps
        prev_1 = 2  # ways to reach step 2
        prev_2 = 1  # ways to reach step 1
        
        current = 0
        for i in range(3, n + 1):
            current = prev_1 + prev_2
            prev_2 = prev_1
            prev_1 = current
            
        return current
```

## 2. Min Cost Climbing Stairs

**Difficulty:** Easy | **LeetCode:** [LC0746](https://leetcode.com/problems/min-cost-climbing-stairs) | [back to top](#problem-list)

**Status:** ✅

**Problem**

You are given an array `cost`, where `cost[i]` is what you pay to step off stair `i`. You may start on stair 0 or stair 1, and each move climbs 1 or 2 stairs. The top is one position past the last stair. Return the minimum total cost to reach the top.

**Approach**

Both solutions use the same 1-D DP idea with optimal substructure: each stair can only be reached from the previous one or the one before it, so its best cost is built from those two. They differ in direction.

**Solution 1: forward DP, two variables**

- `dp[i]` = minimum cost to arrive at stair `i`, having paid for it: `dp[i] = cost[i] + min(dp[i-1], dp[i-2])`.
- Base cases: `dp[0] = cost[0]` and `dp[1] = cost[1]`, since you can start on either stair.
- The top is reachable from either of the last two stairs, so the answer is `min(dp[n-1], dp[n-2])`.
- Only the previous two values are needed, so the table is replaced by `step_1` and `step_2`. The input is left untouched.

**Solution 2: backward DP, in place**

- Reverse the direction: `cost[i] += min(cost[i+1], cost[i+2])`, going from `n-3` down to 0. After the update, `cost[i]` = cheapest cost to finish starting from stair `i`.
- Base cases are the last two stairs, left as they are, since one move from either reaches the top. The answer is `min(cost[0], cost[1])`, since you can start on either.
- It overwrites the input array, which is fine on LeetCode but a side effect to avoid when the caller still needs the original.

**Takeaway:** the two views are "cost to get here" (forward) and "cost to finish from here" (backward). Both are standard in DP, and the same recurrence works in either direction.

**Complexity (both):** time O(n) · space O(1) extra

```python
class Solution1:
    def minCostClimbingStairs(self, arr: List[int]) -> int:
        n = len(arr)

        step_1 = arr[1]
        step_2 = arr[0]

        for i in range(2, n):
            current_cost = arr[i] + min(step_1, step_2)
            step_2 = step_1
            step_1 = current_cost
        
        return min(step_1, step_2)
```

```python
class Solution2:
    def minCostClimbingStairs(self, cost: List[int]) -> int:
        for i in range(len(cost) - 3, -1, -1):
            cost[i] += min(cost[i + 1], cost[i + 2])

        return min(cost[0], cost[1])
```

## 3. N-th Tribonacci Number

**Difficulty:** Easy | **LeetCode:** [LC1137](https://leetcode.com/problems/n-th-tribonacci-number) | [back to top](#problem-list)

**Status:** ✅

**📌 Named sequence: Tribonacci**, the three-term generalisation of **Fibonacci**. Each term is the sum of the previous *three* (0, 1, 1, 2, 4, 7, 13, 24, ...). Fibonacci sums the previous two (0, 1, 1, 2, 3, 5, 8, ...), as in Climbing Stairs.

**Problem**

The Tribonacci sequence is defined by `T0 = 0`, `T1 = 1`, `T2 = 1`, and `T(n+3) = T(n) + T(n+1) + T(n+2)`. Given `n`, return `T(n)`.

**Approach**

Bottom-up DP on a linear recurrence of order 3: `dp[i] = dp[i-1] + dp[i-2] + dp[i-3]`. Plain recursion would recompute the same terms exponentially many times (overlapping subproblems), so filling from the base cases upward is what makes it linear.

**Solution 1: table**

- Base cases set explicitly: `T0 = 0`, `T1 = T2 = 1`. The loop starts at 3, so `dp[i-3]` is never a negative index. In Python that would silently wrap around.
- Time O(n) · space O(n).

**Solution 2: three variables**

- Each term needs only the last three, so the table is replaced by `a, b, c` and updated with `a, b, c = b, c, a + b + c`.
- Time O(n) · space O(1).

**Takeaway:** Fibonacci is the order-2 case and Tribonacci the order-3 case. The same pattern works for any order k, with a window of k variables.

```python
class Solution:
    def tribonacci(self, n: int) -> int:
        if n == 0:
            return 0
        if n <= 2:
            return 1

        dp = [0] * (n + 1)   # dp[i] = T(i)
        dp[1] = dp[2] = 1    # base cases: T0 = 0 (already), T1 = T2 = 1

        for i in range(3, n + 1):
            dp[i] = dp[i - 1] + dp[i - 2] + dp[i - 3]

        return dp[n]
```

```python
class Solution:
    def tribonacci(self, n: int) -> int:
        if n == 0:
            return 0
        a, b, c = 0, 1, 1        # T(i-3), T(i-2), T(i-1)
        for _ in range(3, n + 1):
            a, b, c = b, c, a + b + c
        return c
```

## 4. House Robber

**Difficulty:** Medium | **LeetCode:** [LC0198](https://leetcode.com/problems/house-robber) | [back to top](#problem-list)

**Status:** ✅

**Problem**

Houses are in a row, and `nums[i]` is the money in house `i`. You cannot rob two adjacent houses (that triggers an alarm). Return the maximum total you can rob.

**Approach**

- Define `dp[i]` = best total using houses `0..i`. At house `i` there are two choices: skip it and keep `dp[i-1]`, or rob it and add `nums[i]` to `dp[i-2]`, because house `i-1` is then off limits. So `dp[i] = max(dp[i-1], dp[i-2] + nums[i])`.
- Base cases: `dp[0] = nums[0]` and `dp[1] = max(nums[0], nums[1])`. With one house you take it, and with two you take the larger.
- Theory: 1-D DP with optimal substructure. The "no two adjacent" constraint is what makes `dp[i-2]` the right state to build on.
- Each state depends only on the previous two, so the table is replaced by two variables: `prev_1` = `dp[i-1]` and `prev_2` = `dp[i-2]`.
- The `n == 1` and `n == 2` early returns handle the arrays too short for the loop.

**Complexity:** time O(n) · space O(1)

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        current_money = 0
        if n == 1:
            return nums[0]
        prev_1 = max(nums[0], nums[1])
        prev_2 = nums[0]
        if n == 2:
            return max(prev_1, prev_2)
        for i in range(2,n):
            current_money = max(prev_1, prev_2 + nums[i])
            prev_2 = prev_1
            prev_1 = current_money
        return current_money
```

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        prev_2, prev_1 = 0, 0          # best with 0 houses, best with "-1" houses
        for x in nums:
            prev_2, prev_1 = prev_1, max(prev_1, prev_2 + x)
        return prev_1
```

## 5. House Robber II

**Difficulty:** Medium | **LeetCode:** [LC0213](https://leetcode.com/problems/house-robber-ii) | [back to top](#problem-list)
 
**Status:** ✅
 
**Problem**
 
Same as House Robber, except the houses are arranged in a **circle**, so the first and last houses are neighbors. Return the maximum total you can rob without robbing two adjacent houses.
 
**Approach**
 
The circle breaks the House Robber state: "best on the first `i` houses" does not record whether house 1 was robbed, and that decides whether the last house is legal. The fix is to split into two ordinary line problems.
 
- Houses 1 and `n` are adjacent, so no valid plan robs both.
- Every valid plan therefore either **skips house `n`** (a valid plan on `nums[:-1]`) or **skips house 1** (a valid plan on `nums[1:]`).
- Both are plain line problems, solved by the House Robber DP. The answer is the max of the two.
**Solution: run the linear solver twice**
 
- Reuse `rob_line` from House Robber: `max(rob_line(nums[:-1]), rob_line(nums[1:]))`.
- Special case: `n = 1`. Both slices are empty and would return 0, but the answer is `nums[0]`.
- Cases `n = 2` and `n = 3` need no special handling, since the general formula already gives the right result (the slices have one or two houses).
- Worked examples: `[2, 3, 2]` gives `max(3, 3) = 3` (the two 2's are neighbors in the circle). `[1, 2, 3, 1]` gives `max(4, 3) = 4`.
**Why the overlap is harmless:** a plan that skips both house 1 and house `n` appears in both sub-problems. That does not matter, because we take a **max**, and a candidate counted twice cannot change a max. What matters is that every valid plan is covered (exhaustive), and that everything the sub-solvers return is valid in the circle (a plan on `nums[:-1]` never touches house `n`, so the wrap-around adjacency cannot be violated).
 
| Combining with | Cases must be |
|---|---|
| `+` (counting, e.g. Climbing Stairs) | exhaustive **and** disjoint |
| `max` / `min` (optimizing) | exhaustive only |
 
**Trap: two different graphs.** The circle changes the *adjacency* graph of the houses (a path becomes a cycle). It does **not** make the *subproblem dependency* graph cyclic, which is always a DAG in DP. The linear recurrence would still run on a circle and terminate, but it would give wrong answers, because the state is insufficient, not because of a cycle in the dependencies.
 
**Takeaway:** when a constraint breaks the state definition, split the problem into cases where the constraint no longer applies, and reuse the solver you already have.
 
**Complexity:** time O(n) · space O(1) extra

```python
class Solution:
    def rob_line(nums: list[int]) -> int:
        prev2 = prev1 = 0                      # g(i-2), g(i-1); g(0) = 0
        for x in nums:
            prev2, prev1 = prev1, max(prev1, prev2 + x)
        return prev1


    def rob_circle(nums: list[int]) -> int:
        if len(nums) == 1:                     # both slices would be empty
            return max(0, nums[0])
        return max(rob_line(nums[:-1]), rob_line(nums[1:]))
```

## 9. Coin Change

**Difficulty:** Medium &nbsp;|&nbsp; **LeetCode:** [LC0322](https://leetcode.com/problems/coin-change) &nbsp;|&nbsp; [back to top](#problem-list)

**Status:** ✅

**Problem**

You are given an array `coins` of coin values (each can be used any number of times) and an integer `amount`. Return the fewest coins needed to make exactly `amount`, or `-1` if it cannot be made.

**Approach**

Condition on the **last decision**: the last coin added was some coin `c`. Before it, the remaining amount `a - c` was made with the fewest coins. So try every coin, take the best result for the smaller amount, add 1 for the coin, and keep the minimum. Same skill as Climbing Stairs and Min Cost Climbing, with "which coin was last" as the decision.

**Solution: bottom-up 1-D DP**

- **State:** `dp[a]` = fewest coins to make exactly `a`.
- **Recurrence:** `dp[a] = 1 + min(dp[a - c])` over all coins `c <= a`.
- **Base case:** `dp[0] = 0` (no coins make amount 0).
- **Order:** fill `a = 1, 2, ..., amount`, so `dp[a - c]` is always ready.
- **Answer:** `dp[amount]`, or `-1` if it is still infinity.
- **Impossible amounts:** initialize every cell except `dp[0]` to `float('inf')`. Then `1 + inf = inf` and `min(inf, x) = x`, so impossibility passes through the recurrence by itself. Translate to `-1` only once, at the end.
- Worked example: coins `[1, 3, 4]`, amount 6 gives `dp = 0, 1, 2, 1, 1, 2, 2`, so the answer is **2** (3 + 3).

**Alternative view: BFS on amounts**

- Treat each amount as a node, with an edge from `a` to `a + c` for every coin. The fewest coins is the shortest path (in edges) from 0 to `amount`, which BFS finds layer by layer.
- Same complexity. This is the bridge to graph DP later.

**Trap: common bugs**

- Greedy fails. "Take the biggest coin that fits" gives `4 + 1 + 1 = 3` coins for amount 6 with `[1, 3, 4]`, but the optimum is `3 + 3 = 2`. A local choice changes what remains, so DP tries every coin at every state. (Greedy works only for special coin systems, such as US coins.)
- Initializing the whole table with `0`: `min` can never improve it, so nothing is stored.
- Forgetting the `+ 1`: the table then counts the wrong thing.
- Using `-1` inside the table: `1 + (-1) = 0` looks like a valid answer. Use infinity and convert once at the end.
- Reading `dp[i - c]` with `c > i`: guard with `i >= c`.
- `range(1, amount)` stops one short; it must be `range(1, amount + 1)`.

**Takeaway:** this is **unbounded** knapsack, because `dp[a - c]` may already contain coin `c`, so each coin can be reused. 0/1 Knapsack, where each item is used once, needs a second index. Note for the counting variant (Coin Change II, number of ways): the loop order decides whether you count combinations (coins outer) or ordered sequences (amount outer, as in Climbing Stairs). The table cannot be shrunk to a few variables, because coins reach back arbitrarily far.

**Complexity:** time O(amount · k) for k coins · space O(amount)

```python
class Solution:
    def coin_change(coins: list[int], amount: int) -> int:
        dp = [float('inf')] * (amount+1)
        dp[0] = 0
        for i in range(1, amount+1):
            for c in coins:
                if i >= c:
                    dp[i] = min(dp[i], 1+dp[i-c])
        if dp[amount] == float('inf'):
            return -1
        else:
            return dp[amount]
```

## 9.2 Coin Change II

**Difficulty:** Medium &nbsp;|&nbsp; **LeetCode:** [LC0518](https://leetcode.com/problems/coin-change-ii) &nbsp;|&nbsp; [back to top](#problem-list)

**Status:** ✅

**Problem**

You are given an integer `amount` and an array `coins` of coin values (each can be used any number of times). Return the number of **combinations** of coins that sum to exactly `amount`. Order does not matter: `1 + 2` and `2 + 1` are the same combination. If no combination works, return 0.

**Approach**

Same coins and same unlimited supply as Coin Change, but the question changes from "fewest coins" (optimize, combine with `min`) to "how many ways" (count, combine with `+`). The new difficulty is **not overcounting**: `1 + 2` and `2 + 1` must count once. The fix is to process the coins **one at a time** and fix an order in which coins may appear, so every combination is built in exactly one way.

**Solution: bottom-up 1-D DP, coins in the outer loop**

- **State:** after processing the first `j` coins, `dp[a]` = number of combinations of those coins that make exactly `a`. The single array is updated in place as `j` grows.
- **Recurrence:** for each coin `c`, `dp[a] += dp[a - c]` for `a >= c`. Meaning: combinations of `a` that use coin `c` at least once equal combinations of `a - c` using the coins allowed so far (including `c` again, so it is reusable).
- **Base case:** `dp[0] = 1` (the empty combination counts as one way). It is **1, not 0**: with 0 everything stays 0.
- **Order:** outer loop over coins, inner loop over amounts **increasing** (increasing lets a coin be reused).

**Why the loop order matters**

- **Coins outer, amounts inner:** each combination is generated in exactly one order (all uses of the first coin, then the second, and so on), so it counts **combinations**.
- **Amounts outer, coins inner:** every coin can be the last one at each step, so `1 + 2` and `2 + 1` are both counted. This counts **ordered sequences** (permutations), which is exactly the Climbing Stairs recurrence.
- Small check with `amount = 3`, coins `[1, 2]`: combinations = 2 (`1+1+1`, `1+2`); ordered sequences = 3 (`1+1+1`, `1+2`, `2+1`).

**Trap: common bugs**

- `dp[0] = 0` (wrong): every cell stays 0. The empty combination is one way.
- Swapping the loops: you silently get the permutation count instead, and the example above would return 3, not 2.
- Using `=` instead of `+=`: you overwrite the earlier coins' contribution instead of adding to it.
- Forgetting the guard `a >= c` (or starting the inner loop at `c`): negative index.

**Takeaway:** same table as Coin Change, but `+` replaces `min`, and the loop order now carries meaning. For 0/1 Knapsack (each item used once) the inner loop over amounts goes **downward**, so an item cannot be reused. Unbounded (this problem) goes upward.

**Complexity:** time O(amount · k) for k coins · space O(amount)

```python
class Solution:
    def change(amount, coins):
        dp = [0] * (amount + 1)
        dp[0] = 1
        for c in coins:
            for a in range(c, amount + 1):
                dp[a] += dp[a - c]
        return dp[amount]
```

## 13. Partition Equal Subset Sum

**Difficulty:** Medium &nbsp;|&nbsp; **LeetCode:** [LC0416](https://leetcode.com/problems/partition-equal-subset-sum) &nbsp;|&nbsp; [back to top](#problem-list)

**Status:** ⬜

**Approach**

## 14. Combination Sum IV

**Difficulty:** Medium | **LeetCode:** [LC0377](https://leetcode.com/problems/combination-sum-iv) | [back to top](#problem-list)

**Status:** ✅

**📌 Named problem: unbounded knapsack (counting variant, ordered).** Each item can be used any number of times, and you count the ways to reach an exact total. Here different orderings count as different ways.

**Problem**

Given an array of distinct positive integers `nums` and a `target`, return the number of sequences of numbers from `nums` (each usable any number of times) that add up to `target`. Sequences with the same numbers in a different order count separately, so `(1, 2)` and `(2, 1)` are two ways.

**Approach**

- `dp[i]` = number of ordered sequences that sum to `i`. The last number in a sequence is some `c` in `nums`, and what comes before it is any sequence summing to `i - c`. So `dp[i] = sum of dp[i - c]` over every `c <= i`.
- Base case: `dp[0] = 1`, the empty sequence, so a number equal to `i` contributes exactly one way.
- **Loop order matters:** the outer loop runs over the **totals** `i`, the inner loop over the **numbers** `c`. At every total, every number is tried as the last element, which counts each ordering separately (permutations).
- Theory: 1-D DP with optimal substructure. Swapping the two loops would count combinations instead, which is Coin Change II.

**Complexity:** time O(target · len(nums)) · space O(target)

```python
class Solution:
    def combinationSum4(self, nums: List[int], target: int) -> int:
        dp = [0]*(target+1)
        dp[0] = 1

        for i in range(1, target+1):
            for c in nums:
                if i >= c:
                    dp[i] = dp[i] + dp[i-c]
        return dp[target]
```

## 15. Perfect Squares

**Difficulty:** Medium | **LeetCode:** [LC0279](https://leetcode.com/problems/perfect-squares) | [back to top](#problem-list)

**Status:** ✅

**📌 Named problem: unbounded knapsack (minimum-count variant).** Each item can be reused any number of times, and you minimise how many items reach an exact total. This is the same pattern as Coin Change, with the perfect squares as the "coins".

**Problem**

Given a positive integer `n`, return the smallest number of perfect squares (1, 4, 9, 16, ...) that add up to `n`. The same square can be used more than once.

**Approach**

- Reduce it to Coin Change: the coins are all perfect squares up to `n`, the amount is `n`, and the answer is the minimum number of coins.
- `dp[i]` = fewest squares that sum to `i`. The last square used is some coin `c <= i`, so `dp[i] = min(dp[i], 1 + dp[i - c])`.
- Base case: `dp[0] = 0`. Everything else starts at infinity, meaning "not reached yet".
- The answer is always reachable, because the coin `1` is always available, so the `-1` branch for an unreachable amount never runs here. It is only needed for the general Coin Change helper.
- Theory (not used in the code): Lagrange's four-square theorem says every positive integer is a sum of at most 4 squares, so the answer is always between 1 and 4. It also allows an O(√n) math solution that checks the special cases (a perfect square, a sum of two squares, the form 4^a(8b+7), otherwise 3 or 4).

**Complexity:** time O(n · √n), since there are about √n squares and each is tried for every `i` · space O(n)

```python
class Solution:
    def numSquares(self, n: int) -> int:
        
        def coin_change(coins, amount):
            dp = [float('inf')] * (amount+1)
            dp[0] = 0
            for i in range(1, amount+1):
                for c in coins:
                    if i >= c:
                        dp[i] = min(dp[i], 1+dp[i-c])
            if dp[amount] == float('inf'):
                return -1
            else:
                return dp[amount]

            coins = []
            k = 1
            while k * k <= n:
                coins.append(k * k)
                k += 1
                    
        return coin_change(coins, n)
```
