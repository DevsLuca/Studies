## 1. Lemonade Change

**Difficulty:** Easy &nbsp;|&nbsp; **LeetCode:** [LC0860](https://leetcode.com/problems/lemonade-change) &nbsp;|&nbsp; [back to top](#problem-list)

**Status:** ✅

**Problem**

A lemonade stand sells each lemonade for €5. Customers line up in order, buy exactly one lemonade, and pay with a €5, €10 or €20 bill. You start with no change. Given the list of bills in the order they arrive, return `True` if you can give correct change to every customer, otherwise `False`.

**Approach**

- Greedy with a bill counter. Keep how many €5, €10 and €20 bills you hold, and process customers in order.
- €5: no change needed. €10: give back one €5. €20: you owe €15.
- For the €20 case, prefer **€10 + €5** over **three €5**. Greedy choice: €5 bills are the most flexible, because they are needed for every €10 and can also help with €20. €10 bills can only be used for €20 change, so spending them first never hurts, while spending three €5 bills can.
- The €20 bills you receive are never given back as change, so you do not need to track them. Return `False` as soon as you cannot make change.

**Complexity:** time O(n) · space O(1)

```python
class Solution:
    def lemonadeChange(self, bills: List[int]) -> bool:
        wallet = {5:0, 10:0, 20:0}
        for n in bills:
            wallet[n] += 1
            if n == 10:
                if wallet[5] == 0:
                    return False
                    break    
                else:
                    wallet[5] -= 1
            elif n == 20:
                if wallet[10] == 0 and wallet[5] < 3:
                    return False
                    break
                else:
                    if wallet[10] >= 1 and wallet[5] >= 1:
                        wallet[10] -= 1 
                        wallet[5] -= 1
                    elif wallet[5]>=3:
                        wallet[5] -= 3
                    else:
                        return False
                        break
        return True
```

## 2. Maximum Subarray

**Difficulty:** Medium | **LeetCode:** [LC0053](https://leetcode.com/problems/maximum-subarray) | [back to top](#problem-list)

**Status:** ✅

**📌 Kadane's algorithm** (1984). A single pass that finds the maximum-sum contiguous subarray by tracking the best sum ending at each position, and restarting whenever the running sum would be worse than starting fresh.

**Problem**

Given an integer array `nums` (values can be negative), find the contiguous, non-empty subarray with the largest sum and return that sum.

**Approach**

- Keep `current_sum` = best sum of a subarray that **ends at the current index**, and `max_sum` = best seen so far.
- At each element, either extend the previous subarray (`current_sum + nums[i]`) or start fresh at `nums[i]`. Take the larger.
- Greedy view: a negative running sum can only hurt later subarrays, so drop it and restart.
- Theory: this is a 1-D DP, `best_ending_at[i] = max(best_ending_at[i-1] + nums[i], nums[i])`, compressed to one variable because only the previous state is needed.
- Initialising with `nums[0]` (not 0) handles all-negative arrays, since the subarray must be non-empty.

**Complexity:** time O(n) · space O(1)

```python
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:

        current_sum = nums[0]
        max_sum = nums[0]

        for i in range(1, len(nums)):
            current_sum = max(current_sum + nums[i], nums[i])
            max_sum = max(max_sum, current_sum)

        return max_sum
```
